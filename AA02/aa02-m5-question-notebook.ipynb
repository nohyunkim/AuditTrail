````python
# (코드 셀) 코드의 빈칸을 채우고 실행하세요

q1 = df['FE_Repeat_Call'].quantile(0.25)
q3 = df['FE_Repeat_Call'].quantile(0.75)
iqr = q3 - q1

lower_fence = q1 - 2.0 * iqr
upper_fence = q3 + 2.0 * iqr

df_temp = df.drop( df[(df['FE_Repeat_Call'] > upper_fence) | (df['FE_Repeat_Call'] < lower_fence)].index, axis=0)
df_temp = df_temp.drop(columns=['E_Serial_No'])
df_temp.reset_index(drop=True, inplace=True)
````
