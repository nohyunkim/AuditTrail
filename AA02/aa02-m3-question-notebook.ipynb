### **5. 일 관수량(L)(Water_L) 컬럼의 이상치를 IQR 기준으로 제거하고자 합니다. 제시된 가이드에 맞춰 처리하십시오.**

- 앞 문항 결과 사용 : 2번 `df`
- 대상 데이터프레임 : df
- 사분위수(Q1, Q3)와 IQR 을 계산하세요
- lower_fence = Q1 - 2.0 x IQR, upper_fence = Q3 + 2.0 x IQR 로 정의하세요
- Water_L 이(가) 두 fence 밖에 있는 행을 삭제하세요
- 식별자 컬럼 Uid 도 삭제하고 인덱스를 0부터 다시 매겨 df_temp 에 저장하세요
- 아래 코드 셀(cell)의 코드를 실행해 결과를 확인하세요 — 빈칸(____)은 한 곳입니다
- **[정답05]** 아래 코드 셀의 빈칸(____)에 들어갈 내용(함수·인자·모듈 이름)을 입력하세요
- 답안 : **[정답05]** 의 값만 답안 셀에 `정답05 = 값` 형태로 입력하세요 (글자는 따옴표로 감싸기)
- 배점 : 6점

---

````python
# (코드 셀) 코드의 빈칸(____)을 채우고 실행하세요

q1 = df['Water_L'].quantile(0.25)
q3 = df['Water_L'].quantile(0.75)
iqr = q3 - q1

lower_fence = q1 - 2.0 * iqr
upper_fence = q3 + 2.0 * iqr

is_out = (df['Water_L'] > upper_fence) | (df['Water_L'] < lower_fence)
df_temp = df[~is_out].drop(columns=['Uid']).reset_index(drop=True)
````

````python
# (5) 정답을 입력하세요 — `정답05` 에 값만 넣고 실행하세요

정답05 = 'quantile'
````
