### **5. 그래프로 확인한 결과, 잔여 보증기간(개월)(WARRANTY_M) 컬럼에서 이상치가 발견되었습니다. AI모델링할 때 부정적 영향이 없도록 다음 지침에 따라 이상치를 처리하십시오.**

- 앞 문항 결과 사용 : 2번 `data`
- 대상 데이터프레임 : data
- 잔여 보증기간(개월)(WARRANTY_M)이(가) upper fence 값보다 크거나 lower fence 값보다 작은 이상치 행(row)을 삭제하세요 (IQR 기준 K=1.5)
- AI모델링에 불필요한 행 식별번호(ROW_ID) 컬럼을 삭제하세요
- 전처리를 적용한 데이터프레임은 data_temp 라는 변수명으로 담으세요
- 아래 코드 셀(cell)의 코드를 실행해 결과를 확인하세요 — 빈칸(____)은 한 곳입니다
- **[정답05]** 아래 코드 셀의 빈칸(____)에 들어갈 내용(함수·인자·모듈 이름)을 입력하세요

- 배점 : 6점

---

````python
# (코드 셀) 코드의 빈칸을 채우고 실행하세요

q1 = data['WARRANTY_M'].quantile(0.25)
q3 = data['WARRANTY_M'].quantile(0.75)
iqr = q3 - q1

lower_fence = q1 - 1.5 * iqr
upper_fence = q3 + 1.5 * iqr

data_temp = data.drop( data[(data['WARRANTY_M'] > upper_fence) | (data['WARRANTY_M'] < lower_fence)].index, axis=0)
data_temp = data_temp.drop(columns=['ROW_ID'])
data_temp.reset_index(drop=True, inplace=True)
````
