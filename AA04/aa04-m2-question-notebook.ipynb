### **5. 그래프로 확인한 결과, 반경 1km 학교 수(FE_School_Cnt) 컬럼에서 이상치가 발견되었습니다. AI모델링할 때 부정적 영향이 없도록 제시된 가이드에 맞춰 이상치를 처리하세요.**

<sub>학습 목표: 사분위 범위(IQR) 등의 기준으로 이상치를 찾아 처리할 수 있다 | 역량: DP1 이상치 처리 | 기술: pandas(quantile, drop, reset_index) | 난이도: 쉬움 | 예상 시간: 6분</sub>

- 앞 문항 결과 사용 : 2번 `df_raw`
- 대상 데이터프레임 : df_raw
- 반경 1km 학교 수(FE_School_Cnt)이(가) upper fence 값보다 크거나 lower fence 값보다 작은 이상치 행(row)을 삭제하세요 (IQR 기준 K=1.5)
- AI모델링에 불필요한 등록번호(E_Reg_No) 컬럼을 삭제하세요
- 전처리를 적용한 데이터프레임은 df_clean 라는 변수명으로 저장하세요
- 아래 코드 셀(cell)의 코드를 실행해 결과를 확인하세요 — 빈칸(____)은 한 곳입니다
- **[정답05]** 아래 코드 셀의 빈칸(____)에 들어갈 내용(함수·인자·모듈 이름)을 입력하세요
- 답안 : **[정답05]** 의 값만 답안 셀에 `정답05 = 값` 형태로 입력하세요 (글자는 따옴표로 감싸기)
- 배점 : 6점

````python
# (코드 셀) 코드의 빈칸을 채우고 실행하세요

q1 = df_raw['FE_School_Cnt'].quantile(0.25)
q3 = df_raw['FE_School_Cnt'].quantile(0.75)
iqr = q3 - q1

lower_fence = q1 - 1.5 * iqr
upper_fence = q3 + 1.5 * iqr

df_clean = df_raw.drop( df_raw[(df_raw['FE_School_Cnt'] > upper_fence) | (df_raw['FE_School_Cnt'] < lower_fence)].index, axis=0)
df_clean = df_clean.drop(columns=['E_Reg_No'])
df_clean.reset_index(drop=True, inplace=True)
````
