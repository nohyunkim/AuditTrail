### **5. 회전 속도(rpm)(FE_Rpm) 컬럼의 이상치를 IQR 기준으로 제거하려 합니다. 아래 가이드에 따라 처리하십시오.**

- 앞 문항 결과 사용 : 2번 `df_raw`
- 대상 데이터프레임 : df_raw
- 사분위수(Q1, Q3)와 IQR 을 구하세요
- lower_fence = Q1 - 1.5 x IQR, upper_fence = Q3 + 1.5 x IQR 로 정의하세요
- FE_Rpm 이(가) 두 fence 밖에 있는 행을 삭제하세요
- 식별자 컬럼 E_Uid 도 삭제하고 인덱스를 0부터 다시 매겨 df_clean 에 담으세요
- 아래 코드 셀(cell)의 코드를 실행해 결과를 확인하세요 — 빈칸(____)은 한 곳입니다
- **[정답05]** 아래 코드 셀의 빈칸(____)에 들어갈 내용(함수·인자·모듈 이름)을 입력하세요
- 답안 : **[정답05]** 의 값만 답안 셀에 `정답05 = 값` 형태로 입력하세요 (글자는 따옴표로 감싸기)
- 배점 : 6점

---

````python
# (코드 셀) 코드의 빈칸(____)을 채우고 실행하세요

q1 = df_raw['FE_Rpm'].quantile(0.25)
q3 = df_raw['FE_Rpm'].quantile(0.75)
iqr = q3 - q1

lower_fence = q1 - 1.5 * iqr
upper_fence = q3 + 1.5 * iqr

is_out = (df_raw['FE_Rpm'] > upper_fence) | (df_raw['FE_Rpm'] < lower_fence)
df_clean = df_raw[~is_out].drop(columns=['E_Uid']).reset_index(drop=True)
````

````python
# (5) 정답을 입력하세요 — `정답05` 에 값만 넣고 실행하세요

정답05 = 'quantile'
````

````python
# (코드 셀) 코드의 빈칸을 채우고 실행하세요

model = Sequential([
    Dense(32, activation='relu', input_dim=X_train.shape[1]),
    Dropout(0.5),
    Dense(16, activation='relu'),
    Dense(1)
])

estop = EarlyStopping(monitor='val_loss', patience=7, restore_best_weights=True)
cb_list = [estop]
ckpt = ModelCheckpoint('best_model.keras', monitor='val_loss', save_best_only=True)
cb_list.append(ckpt)

model.compile(optimizer='adam', loss='mse', metrics=['mae'])

history = model.fit(X_train, y_train,
                    epochs=50, batch_size=32,
                    validation_split=0.2,
                    callbacks=cb_list)
````
